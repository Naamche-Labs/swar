# Swar: clone a Nepali voice (free, Google Colab)

Swar is an open Nepali voice model by Naamche Labs. Drop 20 to 30 seconds of someone talking (a WhatsApp voice note is fine), type Nepali, English or both, and hear it in their voice. Or pick one of our voices, Maya or Aarav.

1. **Runtime → Change runtime type → T4 GPU** (free), then **Runtime → Run all**.
2. After about 5 minutes a link ending in `gradio.live` appears below. Open it.
3. Only clone a voice with that person's permission.

```
# 2. Download the model and the app, then start it
import os, sys
from huggingface_hub import snapshot_download
REPO = "Aarjan/swar"
os.environ["MODEL_REPO"] = REPO
d = snapshot_download(REPO)
os.chdir(f"{d}/app"); sys.path.insert(0, f"{d}/app")
import app
app.demo.queue(default_concurrency_limit=1).launch(share=True, debug=True)
```
